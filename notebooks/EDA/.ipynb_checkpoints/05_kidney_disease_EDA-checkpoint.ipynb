```python
# HealthGuard AI - Kidney Disease EDA
# Author: HealthGuard AI Team
# Date: 2026
# Purpose: Exploratory Data Analysis of Kidney Disease Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

plt.style.use('seaborn-v0_8')
sns.set_palette('Set2')

print("=" * 50)
print("   HealthGuard AI - Kidney Disease EDA")
print("=" * 50)
print("Libraries Loaded Successfully!")
```

```python
# Load Kidney Disease Dataset
# Source: UCI Machine Learning Repository / Kaggle
# Records: 400 patients, 25 columns

df = pd.read_csv("E:/HealthGuard_AI/data/raw/kidney_disease.csv")

print("Dataset Loaded Successfully!")
print("=" * 50)
print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}")
print("=" * 50)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nFirst 5 Rows:")
df.head()
```

```python
# Basic Dataset Information

print("=" * 50)
print("DATASET INFORMATION")
print("=" * 50)
df.info()
```

```python
# Basic Statistical Analysis

print("=" * 50)
print("BASIC STATISTICS")
print("=" * 50)
df.describe().round(2)
```

```python
# Check Missing Values
# Kidney dataset has many missing values

print("=" * 50)
print("MISSING VALUES CHECK")
print("=" * 50)

missing = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_percent.round(1)
})

print(missing_df[missing_df['Missing Count'] > 0])
print(f"\nTotal Missing Values: {df.isnull().sum().sum()}")
```

```python
# Check Duplicate Rows

print("=" * 50)
print("DUPLICATE CHECK")
print("=" * 50)

duplicates = df.duplicated().sum()
print(f"Total Duplicate Rows: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print(f"Duplicates Removed!")
    print(f"New Shape: {df.shape}")
else:
    print("No Duplicates Found!")
```

```python
# Target Variable Distribution
# Fix classification column first

print("=" * 50)
print("TARGET DISTRIBUTION")
print("=" * 50)

# Clean classification column - remove extra spaces/tabs
df['classification'] = df['classification'].str.strip()

print("Kidney Disease Distribution:")
print(df['classification'].value_counts())

percentage = df['classification'].value_counts(normalize=True) * 100
print(f"\nCKD: {percentage['ckd']:.1f}%")
print(f"Normal: {percentage['notckd']:.1f}%")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

values = df['classification'].value_counts()

axes[0].bar(['Kidney Disease', 'Normal'],
            [values['ckd'], values['notckd']],
            color=['#e74c3c', '#2ecc71'],
            edgecolor='black')
axes[0].set_title('Kidney Disease Distribution - Count')
axes[0].set_ylabel('Count')
for i, v in enumerate([values['ckd'], values['notckd']]):
    axes[0].text(i, v + 1, str(v), ha='center', fontweight='bold')

axes[1].pie([values['ckd'], values['notckd']],
            labels=['Kidney Disease', 'Normal'],
            autopct='%1.1f%%',
            colors=['#e74c3c', '#2ecc71'],
            startangle=90)
axes[1].set_title('Kidney Disease Distribution - Percentage')

plt.suptitle('Target Variable Distribution - Kidney Disease',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("Target Distribution Complete!")
```

```python
# Numeric Feature Distributions

numeric_cols = df.select_dtypes(
    include=['float64', 'int64']).columns.tolist()

fig, axes = plt.subplots(4, 3, figsize=(16, 16))
axes = axes.flatten()

colors = ['#3498db', '#e74c3c', '#2ecc71', '#f39c12',
          '#9b59b6', '#1abc9c', '#e67e22', '#34495e',
          '#e91e63', '#00bcd4', '#8bc34a', '#ff5722']

for i, col in enumerate(numeric_cols[:12]):
    axes[i].hist(df[col].dropna(), bins=20,
                 color=colors[i],
                 edgecolor='black',
                 alpha=0.7)
    axes[i].set_title(f'Distribution of {col}',
                      fontweight='bold')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')

plt.suptitle('Numeric Feature Distributions - Kidney Disease',
             fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

print("Distribution Analysis Complete!")
```

```python
# Correlation Heatmap

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df_encoded = df.copy()
cat_cols = df_encoded.select_dtypes(include=['object']).columns
for col in cat_cols:
    df_encoded[col] = le.fit_transform(
        df_encoded[col].astype(str))

plt.figure(figsize=(20, 16))
correlation = df_encoded.corr(numeric_only=True)

mask = np.triu(np.ones_like(correlation, dtype=bool))

sns.heatmap(correlation,
            annot=True,
            fmt='.2f',
            cmap='coolwarm',
            linewidths=0.5,
            mask=mask,
            annot_kws={'size': 8})

plt.title('Correlation Heatmap - Kidney Disease Dataset',
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("Top Correlations with Kidney Disease:")
print(correlation['classification'].sort_values(ascending=False))
```
